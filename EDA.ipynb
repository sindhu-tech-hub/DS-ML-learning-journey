```python
#handling nulls
#Fill them with specific data or suitable data like mean, median, mode, ffill, bfill
#drop them if the number of nulls are less
```

```python
df.dropna(subset=["product"],inplace=True)
#we dont know the name itself, whats the use of having the other columns? so we dropped it
```

```python
#whichever rows have the brand as null, we're filling with unknown
df.fillna({'brand':'Unknown'},inplace=True)
#or
#df["brand"]=df["brand"].fillna("Unknown")
```

```python
#filling the description with something
df["description"]=df["description"].fillna("No description")
```

```python
#checking
print(df[df["description"]=="No description"])
```

```python
#handling nulls in rating

#filling the nulls with the mean of the entire dataset is not very ideal
#this is because we can't consider butter ratings for hairoil
#or, we cant consider one brand rating to calculate the ratings for another brand or seller

#hence, we categorise based on brand or item type to fill ratings with more logical values
#they shouldn't be misleading or incorrectly estimated
```

```python
#so we categorize based on all -> company, category, sub-category
#so each group has similar products, the null values in each group can be filled with
#mean from that group only
```

```python
# TRANSFORM Function is used to repeatedly find means for all the categories/groups
```

```python
# transform() takes a function and applies it to each column/group, then returns
# a result with the same number of rows as the original data.
```

```python
#https://chatgpt.com/share/6a17f803-5b80-8321-928b-dccf44eb7073
#About groupby and transform
```

```python
df['rating']=df.groupby(['brand','category','sub_category'])['rating'].transform(lambda x:x.fillna(x.mean()))

#check for nulls again
df.isna().sum()
#3591 nulls -> there are groups that have the same brand, cat and sub cat put together
# that have null ratings
```

```python
df1=pd.DataFrame(df.groupby(['brand','category','sub_category'])['rating'])
```

```python
df1.iloc[3,1]
```

```python
# the above group when filtered by brand, cat and sub cat => that group has all nulls
```

```python
# hence, now make the grouping less specific and try -> group by only brand
```

```python
df['rating']=df.groupby('brand')['rating'].transform(lambda x:x.fillna(x.mean()))
```

```python
#check nulls again
df.isna().sum()
```

```python
# we still have nulls -> now we exhausted all categorizations so fill the ratings of the mean of all the ratings now
```

```python
df['rating']=df['rating'].fillna(df['rating'].mean())
df.isna().sum()
#no nulls
```
