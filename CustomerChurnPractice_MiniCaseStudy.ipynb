Build a simple linear model where dependent varaible is 'MonthlyCharges' and independent varaible is tenure.

Divide the dataset into train and test set in 70:30 ratio.

Build the model on train set and predict the values on test set c, after predicting the values, find the root mean square error.

Find out the error in prediction & store the result in 'error'.
Find out the root mean square error.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
model=LinearRegression()
x=pd.DataFrame(df['tenure'])
y=df['MonthlyCharges']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
x
#X = data['tenure'].values.reshape(-1,1)
#-1 -> automatically figures out the number of rows required
# 1-> we need only 1 column
#.values converts into a 1-D array and x needs input as 2-D(model needs)
#hence, we are reshaping
```

```python
model.fit(x_train,y_train)
```

```python
y_pre=model.predict(x_test)
y_pre
```

```python
from sklearn.metrics import r2_score,mean_squared_error
r2_score(y_test,y_pre)
```

```python
mse = mean_squared_error(y_test,y_pre)
rmse = np.sqrt(mse)
print("Root mean square error:",rmse)
```
