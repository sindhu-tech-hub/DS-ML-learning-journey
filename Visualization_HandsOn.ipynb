4.Create pair plots for the entire data to study various patterns in the data.

a.Create pair plots with respect to brand, number of cylinders, year, etc.

Pair of Area plot and Scatter plot

```python
plt.figure(figsize=(25000,100))
sb.pairplot(df,hue="brand")
plt.show()
#pairplot - every numeric column vs every other numeric column
# numeric vs numeric relationship => scatterplot
# univariate chart(same variable on x and y)
# univariate analysis is basically analyzing the frequency of each data point
#if data points are categorical -> bar chart
#data points are continuous -> histogram or box plots if we want to see the central tendency
```


5.Create a heatmap for the entire data to study correlation between each of the columns.

```python
# A heatmap is a pairplot in which each subplot is a tile representing the correlation
# corelation of every numeric column vs every other numeric column
# it cannot extract numeric columns only unlike the pairplot, so we give manually

x=df.iloc[:,:-1] # leaving the last column brand which is of object type
sb.heatmap(x.corr(),annot=True)
#annot gives the value of the corelation constant.
```

```python
# 1. Create the matrix of averages
df_avg = df.groupby("brand")[["hp", "time-to-60", "mpg", "weightlbs"]].mean()

# 2. Pass it to the heatmap
sb.heatmap(df_avg, annot=True,fmt=".1f") #fmt-> format
```

heatmaps can be used like the above too
