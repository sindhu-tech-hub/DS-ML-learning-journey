#Dataset Overview

This is a telecom customer churn dataset used to predict whether a customer will leave the company (Churn = Yes) or stay (Churn = No).

The dataset contains:

Customer demographics
Service subscriptions
Account information
Billing and payment details
Churn status (target variable)

21. Churn:	Whether the customer left the company (target variable)

#Model Building

```python
x=df.drop('Churn',axis=1)
x
```

```python
y=df['Churn']
```

```python
from sklearn.model_selection import train_test_split
```

```python
y.value_counts()
#there is somewhat imbalance -> the model may train only on one category only => use stratify
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
#based on the imbalances in the y column, make the selection
```

```python
from random import Random #this is used for random selection -> not compulsory
from sklearn.ensemble import RandomForestClassifier
```

```python
model=RandomForestClassifier(n_estimators=100)  #hyperparameter for setting the number of decision trees
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_pred,y_test)*100
```
