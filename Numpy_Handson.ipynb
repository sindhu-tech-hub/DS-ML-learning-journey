**Create a random array, with the range 0-1000, with a size 50.**

Perform the following operations on the same.

a. Print the first ten elements of the array.

b. Print the last ten elements of the array

c. Print the elements from the indexes 10-25

d. Print the element at the index 22

e. Print the array using negative indexing.

f. Print the last ten elements of the array, using negative indexing

```python
import numpy as np

# Create random array of size 50 from those values
arr = np.random.randint(0,1000, size=50)

print("Random Array:\n", arr)

# a. First ten elements
print("\na. First ten elements:")
print(arr[ : 10])   # start index = 0 , end index + 1 = 10

# b. Last ten elements
print("\nb. Last ten elements:")
print(arr[-10:])

# c. Elements from indexes 10–25
print("\nc. Elements from index 10 to 25:")
print(arr[10:26])   # 26 because upper bound is exclusive

# d. Element at index 22
print("\nd. Element at index 22:")
print(arr[22])

# e. Entire array using negative indexing (reverse)
print("\ne. Array using negative indexing (reversed):")
print(arr[::-1])

# f. Last ten elements using negative indexing
print("\nf. Last ten elements (negative indexing):")
print(arr[-10:])
```

**Given a numpy array that has values = [1,2,3,4,5]. Perform the following operations.**

a. Insert the element 6 at the end of the array.

b. Insert the element 10 at the beginning of the array.

c. Insert the element 30, before 4.

d. Update the element 5 as 50.

e. Delete the element 50 in the updated array.

f. Create a function to search for the element 30.

g. Sort the array in ascending order

```python
arr = np.array([1,2,3,4,5])
print(arr)
print()

# a. Insert the element 6 at the end of the array.
np.append(arr,6)
print(arr)

#b. Insert the element 10 at the beginning of the array.
np.insert(arr,0,10)
print(arr)

# c. Insert the element 30, before 4.
np.insert(arr,4,30)
print(arr)

#d. Update the element 5 as 50.
arr[4] = 50
print(arr)

# e. Delete the element 50 in the updated array.
np.delete(arr,4)
print(arr)

# f. Create a function to search for the element 30.
def search(arr,x):
    for i in range(len(arr)):  # One by one we are checking the element lik index 0 , index 1 , index 2 , index 3
        if arr[i] == x:
            return i

search(arr,3)

# g. Sort the array in ascending order
np.sort(arr)
print(arr)
```
