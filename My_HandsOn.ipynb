***To convert object or string into date***


```python
d={
    "Date":["10/05/2025","24/05/2026","18/11/2024"] # format-> %d/%m/%Y -> Y(Long year 2015,2025), y(short yr 25,15)
      # Days/Month/Year -> %d/%m/%Y
    }
#a user defined structure containing dates as strings

data=pd.DataFrame(d)
#converting the structure to data frame

data["Date"]=pd.to_datetime(data["Date"],format="%d/%m/%Y")
# there is only one column in the data frame => date
#assigning the formatted column to the same column again
#to_datetime(what to convert,format) [FUNCTION TO CONVERT INTO DATE_TIME]
# y vs Y [long year vs short year]
print(data.dtypes)
# Extracting years from Date
print("Years=\n",data["Date"].dt.year)
# Extracting Month
print("Month=\n",data["Date"].dt.month)
# Days
print("Days=\n",data["Date"].dt.day)


# Creating new column->
# data["Year"]=data["Date"].dt.year
```

3) Convert the Order Date and Ship Date to datetime
Create a new colum: Order_year, Order_month
Create a column Order Delay-> Calculate the shipping delay
(Ship Date - Order date) in days.

```python
df["Order Date"]=pd.to_datetime(df["Order Date"],format="%d/%m/%Y")
df["Ship Date"]=pd.to_datetime(df["Ship Date"],format="%d/%m/%Y")
df["Order Year"]=df["Order Date"].dt.year
df["Order Month"]=df["Order Date"].dt.month
df["delay"]=(df["Ship Date"]-df["Order Date"]).dt.days
df
#SEPARATOR SHOULD BE GIVEN IN THE FORMAT
```
